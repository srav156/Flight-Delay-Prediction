```python
reg_features_train = pd.read_csv("Data/Regression3/reg_features_train.csv", index_col=0)
reg_features_test  = pd.read_csv("Data/Regression3/reg_features_test.csv",  index_col=0)
reg_labels_train   = pd.read_csv("Data/Regression3/reg_labels_train.csv")["ArrDelayMinutes"].to_numpy()
reg_labels_test    = pd.read_csv("Data/Regression3/reg_labels_test.csv")["ArrDelayMinutes"].to_numpy()
```

```python
perf_df = pd.DataFrame(columns=["Regressors", "MSE", "RMSE", "MAE", "R2"])
def print_metrics(reg_labels_test, model_pred, regressor_name, perf_df):
    
    mse = mean_squared_error(reg_labels_test, model_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(reg_labels_test, model_pred)
    r2 = r2_score(reg_labels_test, model_pred)
    
    print(f"MSE      : {mse}", end="\n\n")
    print(f"RMSE     : {rmse}", end="\n\n")
    print(f"MAE      : {mae}", end="\n\n")
    print(f"R2 Score : {r2}", end="\n\n")
    
    new_row = pd.DataFrame([{
        "Regressors": regressor_name,
        "MSE": mse,
        "RMSE": rmse,
        "MAE": mae,
        "R2": r2,
    }])
    perf_df = pd.concat([perf_df, new_row], ignore_index=True)
    return perf_df
```

```python
#Linear Regressor

# model = LinearRegression(n_jobs=-1)
# model.fit(reg_features_train, reg_labels_train)
# joblib.dump(model, "Data/Regression3/LinearRegression.joblib")
model = joblib.load("Data/Regression3/LinearRegression.joblib")
model_pred = model.predict(reg_features_test)
perf_df = print_metrics(reg_labels_test, model_pred, "LinearRegression", perf_df)
del model
del model_pred
```

```python
#Decision Trees Regressor

# model = DecisionTreeRegressor()
# model.fit(reg_features_train, reg_labels_train)
# joblib.dump(model, "Data/Regression3/DecisionTreeRegressor.joblib")
model = joblib.load("Data/Regression3/DecisionTreeRegressor.joblib")
model_pred = model.predict(reg_features_test)
perf_df = print_metrics(reg_labels_test, model_pred, "DecisionTreeRegressor", perf_df)
del model
del model_pred
```

```python
#XGBoost Regressor

# model = XGBRegressor()
# model.fit(reg_features_train, reg_labels_train)
# joblib.dump(model, "Data/Regression3/XGBRegressor.joblib")
model = joblib.load("Data/Regression3/XGBRegressor.joblib")
model_pred = model.predict(reg_features_test)
perf_df = print_metrics(reg_labels_test, model_pred, "XGBRegressor", perf_df)
del model
del model_pred
```
